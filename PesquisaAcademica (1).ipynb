```python
for coluna in dados.columns[13:]:
    sem_consumo = dados[coluna].isin(["CL0", "CL1", "CL2"]).sum()
    com_consumo = dados[coluna].isin(["CL3", "CL4", "CL5", "CL6"]).sum()

    print(f"\n{coluna}")
    print(f"Sem consumo recente: {sem_consumo}")
    print(f"Com consumo recente: {com_consumo}")
```

```python
substancias_selecionadas = [
    "alcool",
    "anfetaminas",
    "poppers",
    "benzos",
    "cannabis",
    "cocaina",
    "ecstasy",
    "heroina",
    "quetamina",
    "legal_high",
    "lsd",
    "metanfetamina",
    "cogumelos",
    "nicotina"
]
```

```python
tracos_bigfive = [
    "neuroticismo",
    "extroversao",
    "abertura_experiencia",
    "amabilidade",
    "conscienciosidade"
]
```

```python
matTracos = dados[tracos_bigfive].copy()
```

```python
matTracos.shape
```

```python
matSub = dados[substancias_selecionadas].apply(
    lambda coluna: coluna.isin(
        ["CL3", "CL4", "CL5", "CL6"]
    ).astype(int)
)
```

```python
matSub.shape
```

```python
matSub.sum()
```

```python
from scipy.stats import pointbiserialr
```

```python
resultados_correlacao = []

for substancia in matSub.columns:
    for traco in matTracos.columns:

        resultado = pointbiserialr(
            matSub[substancia],
            matTracos[traco]
        )

        resultados_correlacao.append({
            "Substancia": substancia,
            "Traco": traco,
            "Correlacao": resultado.statistic,
            "P_valor": resultado.pvalue
        })

correlacoes = pd.DataFrame(resultados_correlacao)
```

```python
correlacoes.head(10)
```

```python
correlacoes.shape
```

```python
significativas = correlacoes[
    correlacoes["P_valor"] < 0.05
].copy()

significativas = significativas.sort_values(
    by="P_valor"
)

significativas
```

```python
significativas_por_traco = (
    significativas["Traco"]
    .value_counts()
)

significativas_por_traco
```

```python
matriz_correlacao = correlacoes.pivot(
    index="Substancia",
    columns="Traco",
    values="Correlacao"
)

matriz_correlacao
```

```python
plt.figure(figsize=(10, 8))

plt.imshow(
    matriz_correlacao,
    cmap="coolwarm",
    aspect="auto",
    vmin=-0.4,
    vmax=0.4
)

plt.colorbar(label="Correlação ponto-bisserial")

plt.xticks(
    range(len(matriz_correlacao.columns)),
    matriz_correlacao.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(matriz_correlacao.index)),
    matriz_correlacao.index
)

plt.title("Correlação entre personalidade e consumo de substâncias")
plt.tight_layout()
plt.show()
```

```python
estatisticas_tracos = matTracos.describe()

estatisticas_tracos
```

```python
comparacao_cannabis = matTracos.groupby(
    matSub["cannabis"]
).mean()

comparacao_cannabis.index = [
    "Sem consumo recente",
    "Com consumo recente"
]

comparacao_cannabis
```

```python
comparacoes = {}

for substancia in matSub.columns:

    comparacoes[substancia] = (
        matTracos.groupby(matSub[substancia]).mean()
    )

comparacoes["cannabis"]
```

```python
len(comparacoes), list(comparacoes.keys())
```

```python
tabelas_comparacao = []

for substancia, tabela in comparacoes.items():

    for traco in matTracos.columns:

        tabelas_comparacao.append({
            "Substancia": substancia,
            "Traco": traco,
            "Media_sem_consumo": tabela.loc[0, traco],
            "Media_com_consumo": tabela.loc[1, traco]
        })

tabela_comparacao = pd.DataFrame(tabelas_comparacao)

tabela_comparacao
```

```python
tabela_comparacao["Diferenca"] = (
    tabela_comparacao["Media_com_consumo"]
    - tabela_comparacao["Media_sem_consumo"]
)
```

```python
maiores_diferencas = tabela_comparacao.nlargest(
    5, "Diferenca"
)

maiores_diferencas
```

```python
menores_diferencas = tabela_comparacao.nsmallest(
    5, "Diferenca"
)

menores_diferencas
```

```python
tabela_comparacao["Diferenca_absoluta"] = (
    tabela_comparacao["Diferenca"].abs()
)

menores_diferencas = tabela_comparacao.nsmallest(
    10, "Diferenca_absoluta"
)

menores_diferencas
```

```python
maiores_diferencas_absolutas = tabela_comparacao.nlargest(
    10, "Diferenca_absoluta"
)

maiores_diferencas_absolutas
```
